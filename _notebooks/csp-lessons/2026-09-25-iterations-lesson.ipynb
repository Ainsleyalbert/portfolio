Popcorn hack: practice with loops by repeating a task and tracking totals.

```python
hours = [3, 5, 2, 4]
total = 0
for h in hours:
    total += h

average = total / len(hours)
print("Total:", total)
print("Average:", average)
```

popcorn hack 2:

```python

alerts = [1, 4, 2, 6, 3, 9, 5]

# Set alert cutoffs
safe_cutoff = 3
critical_cutoff = 9

# Initialize counters
action_count = 0
safe_count = 0
total_action_score = 0
critical_found = False

# Loop through every alert
for alert in alerts:

    # Skip safe alerts
    if alert < safe_cutoff:
        safe_count += 1
        print("Alert " + str(alert) + ": safe, skipping")
        continue

    # Count alerts requiring action
    action_count += 1
    total_action_score += alert
    print("Alert " + str(alert) + ": action needed")

    # Stop at the first critical alert
    if alert >= critical_cutoff:
        critical_found = True
        print("Critical alert found - stopping scan")
        break

# Final summary
print("Safe alerts skipped: " + str(safe_count))
print("Action alerts: " + str(action_count))
print("Total action score: " + str(total_action_score))
print("Critical found: " + str(critical_found))
```

homework hack:

```python

# Wildfire alert levels from different zones
wildfire_alerts = [1, 4, 2, 6, 3, 5, 7, 9, 4, 2]

# Set thresholds
safe_cutoff = 3
critical_cutoff = 9

# Summary variables
checked_count = 0
action_count = 0
skipped_count = 0
total_score = 0
critical_found = False

# Scan wildfire alerts
for alert in wildfire_alerts:

    checked_count += 1

    # Skip safe alerts
    if alert < safe_cutoff:
        skipped_count += 1
        print("Wildfire alert " + str(alert) + ": Safe zone, skipping.")
        continue

    # Count alerts requiring action
    action_count += 1
    total_score += alert
    print("Wildfire alert " + str(alert) + ": Response crew needed!")

    # Stop at a critical alert
    if alert >= critical_cutoff:
        critical_found = True
        print("CRITICAL WILDFIRE ALERT! Dispatching emergency response.")
        break

# Final report
print("\n===== SFSRC WILDFIRE DISPATCH REPORT =====")
print("Zones checked: " + str(checked_count))
print("Safe zones skipped: " + str(skipped_count))
print("Action alerts: " + str(action_count))
print("Total action score: " + str(total_score))
print("Critical alert found: " + str(critical_found))
```
