```python
# Your UESL prototype goes here.
level_stars = [12, 18, 10]

# Add 5 stars to the second level
if len(level_stars) > 1:
    level_stars[1] = level_stars[1] + 5

# Add a new level with 7 stars
level_stars.append(7)

# Print each level's stars
for stars in level_stars:
    print(stars)

# Print the number of levels
print(len(level_stars))

# Calculate the total using traversal
total = 0
for stars in level_stars:
    total = total + stars

print(total)
```
