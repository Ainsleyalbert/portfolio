```python
myHTML = """
<div class="ocs__grid ocs__grid--standard">
  <div class="ocs__grid-cell ocs__grid-cell--header">Temperature Trial (°C)</div>
  <div class="ocs__grid-cell">Beaker 1: 22.4</div>
  <div class="ocs__grid-cell">Beaker 2: 22.9</div>
  <div class="ocs__grid-cell ocs__grid-cell--accent">Beaker 3: 31.6</div>
</div>
<div class="ocs__grid ocs__grid--calculator">
  <div class="ocs__grid-cell">7</div>
  <div class="ocs__grid-cell">8</div>
  <div class="ocs__grid-cell">9</div>
  <div class="ocs__grid-cell ocs__grid-cell--accent">Avg</div>
</div>
"""

class_list = []
import re

for found in re.findall(r'class="([^"]*)"', myHTML):
  class_list.extend(found.split())

made_up = [
    name for name in class_list
    if not name.startswith("ocs__") and not name.startswith("cols-")
]

def line(label, ok):
    return f"<li>{'PASS' if ok else 'FIX'}  {label}</li>"

checks = "<ul>" + "".join([
    line("No inline style attributes", "style=" not in myHTML),
    line(
        "No made up classes" + (
            f" (found: {', '.join(made_up)})" if made_up else ""
        ),
        not made_up
    ),
    line("Has an ocs__grid root", "ocs__grid" in class_list),
    line("Uses the standard variant for Grid 1", "ocs__grid--standard" in class_list),
    line("Uses the calculator variant for Grid 2", "ocs__grid--calculator" in class_list),
    line("Every box uses ocs__grid-cell", class_list.count("ocs__grid-cell") >= 8),
    line("Title is a header cell", "ocs__grid-cell--header" in class_list),
    line(
        "Has accent cells (outlier beaker and Avg key)",
        class_list.count("ocs__grid-cell--accent") >= 2
    ),
]) + "</ul>"

display(HTML(checks + myHTML))
```
