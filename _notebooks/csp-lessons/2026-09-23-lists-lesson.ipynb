popcorn hack 3:

```python
seasons = ["winter", "summer", "fall", "summer", "spring", "summer", "fall"]   # use real class data

frequency = {}
for season in seasons:
    if season in frequency:
        frequency[season] += 1
    else:
        frequency[season] = 1

popular = []
for season in frequency:
    if frequency[season] >= 2:
        popular.append(season)

print(popular)  
```

Homework:

```python

votes = ["chips", "cookies", "chips", "fruit", "cookies", "chips",
         "pretzels", "cookies", "chips", "fruit", "chips", "cookies"]

prices = [2, 3, 2, 4, 3, 2, 1, 3, 2, 4, 2, 3]

print(len(votes))

print(votes[-1])

total = 0

for price in prices:
    total = total + price

average = total / len(prices)

print("Total:", int(total))
print("Average:", int(average))

frequency = {}

for vote in votes:
    if vote in frequency:
        frequency[vote] += 1
    else:
        frequency[vote] = 1

print(frequency)

popular_snacks = []

for snack in frequency:
    if frequency[snack] >= 3:
        popular_snacks.append(snack)

print(popular_snacks)
```
