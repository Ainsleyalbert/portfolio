1: Be a little careful, but this is probably fine.
2: it will never run the outputs of 9 or 6 becuse of the order.

```python
risk_score = 9

if risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")
elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")
elif risk_score >= 3:
    print("Be a little careful, but this is probably fine.")
else:
    print("Nothing suspicious in this message.")
    
```

Popcorn hack 3:

```python
workshops = ["Self-Defense", "Parenting Skills", "Fitness", "Self-Sufficiency"]
signed_up = [12, 7, 15, 9]
capacity = 12

open_workshops = []

for i in range(len(workshops)):
    if signed_up[i] < capacity:
        open_workshops.append(workshops[i])

print("Workshops with open seats:", len(open_workshops))
print(open_workshops)
```

Homework:

```python
message = "URGENT: send gift cards now to unlock your account"
amount_requested = 500
known_contact = False

risk_score = 0

# Add points for each warning sign
if "URGENT" in message:
    risk_score += 2

if "gift cards" in message:
    risk_score += 3

if amount_requested >= 500:
    risk_score += 2

if not known_contact:
    risk_score += 2

# Report the final score
if risk_score >= 7:
    print("High risk")
elif risk_score >= 4:
    print("Medium risk")
else:
    print("Low risk")

print("Risk score:", risk_score)

# 4 thresholds -> 3 outcomes -> 3 answers
```
